```python

#Read the file it_services_tickets_raw.csv into a DataFrame called df.

df = pd.read_csv('E:/AL_ML/CDAC-Training/it_services_tickets_raw.csv')
```

```python
#Change all column names to lowercase, and replace spaces with underscores.
df.alter = df.columns.str.lower().str.replace(" ", "_")
print(df.alter)
```

```python
#Print how many missing values each column has.
print(df.isnull().sum())
```

```python
#Print the total number of missing values in the whole table.
print(df.isnull().sum().sum())
```

```python
#Replace the text values NA, N/A, null and - with proper missing values.
df = df.replace(["NA", "N/A", "null", "-"], pd.NA)
```

```python
#Count how many duplicate rows there are.
print(df.duplicated().sum())
```

```python
#Remove the duplicate rows and print the new number of rows.
df = df.drop_duplicates()
print(len(df))
```

```python
#Show all the tickets where priority is P1.
print(df[df["PRIORITY"] == "P1"])
```

```python
#Show all the tickets where status is Open.
print(df[df["Status"] == "Open"])
```

```python
#Count how many tickets there are for each status.
print(df["Status"].value_counts())
```

```python
#Count how many tickets there are for each priority.
print(df["PRIORITY"].value_counts())
```

```python
#Count how many tickets there are for each service line.
print(df["ServiceLine"].value_counts())
```

```python
#Find how many different clients there are.
print(df["ClientID"].nunique())
```

```python
#Print the list of all countries in the data.
print(df["Client_Country"].unique())
```

```python
#Convert the hours logged column to a number.
df["hours_logged"] = pd.to_numeric(df["hours_logged"], errors="coerce")
print(df["hours_logged"] )
```

```python
#Convert the billing rate column to a number. Remove the $ sign first.
df["billing_rate"] = (
    df["Billable Amount USD"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["billing_rate"] = pd.to_numeric(
    df["billing_rate"],
    errors="coerce"
)
print(df["billing_rate"])
```

```python
#Convert the CSAT score column to a number.
df["csat_score"] = pd.to_numeric(
    df["CSAT Score"],
    errors="coerce"
)
print(df["csat_score"] )
```

```python
#Print the average, the smallest and the largest value of hours logged.
print("Average:", df["hours_logged"].mean())
print("Minimum:", df["hours_logged"].min())
print("Maximum:", df["hours_logged"].max())
```

```python
#Sort the table by hours logged from highest to lowest and show the top 10 rows.
print(
    df.sort_values(
        by="hours_logged",
        ascending=False
    ).head(10)
)
```

```python
#Show the 5 tickets with the highest billable amount.
df["billable_amount"] = pd.to_numeric(
    df["Billable Amount USD"],
    errors="coerce"
)
print(df["billable_amount"].nlargest(5))
```

```python
#Remove the extra spaces from the client name column.
print(df["Client Name "].str.strip())
```

```python
#Change the client name column to title case.
print(df["Client Name "].str.title())
```

```python
#Convert the ticket created column to a date.
print(pd.to_datetime(
    df["Ticket_Created_At"],
    errors="coerce"
))
```

```python
#Add a new column that holds the year of the ticket.
df["Ticket_Created_At"] = pd.to_datetime(
    df["Ticket_Created_At"],
    format="%d-%m-%Y",
    errors="coerce"
)

df["year"] = df["Ticket_Created_At"].dt.year
```

```python
#Add a new column that holds the month of the ticket.
df["month"] = df["Ticket_Created_At"].dt.month

print(df[["Ticket_Created_At", "month"]].head())
```

```python
#Count how many tickets were created in each year.
print(df["year"].value_counts().sort_index())
```

```python
#Add a new column called total equal to billing rate multiplied by hours logged.
df["Billing Rate (USD)"] = pd.to_numeric(
    df["Billing Rate (USD)"],
    errors="coerce"
)

df["hours_logged"] = pd.to_numeric(
    df["hours_logged"],
    errors="coerce"
)

df["total"] = df["Billing Rate (USD)"] * df["hours_logged"]

print(df["total"])
```

```python
#Fill the missing values in the CSAT score column with the average CSAT score.
df["CSAT Score"] = pd.to_numeric(
    df["CSAT Score"],
    errors="coerce"
)

df["CSAT Score"] = df["CSAT Score"].fillna(
    df["CSAT Score"].mean()
)

print(df["CSAT Score"])
```

```python
#Drop the rows where the client name is missing.
df = df.dropna(subset=["Client Name "])
```

```python
#Group by service line and find the average hours logged.
print(
    df.groupby("ServiceLine")["hours_logged"]
    .mean()
)
```

```python
#Group by priority and count the tickets.
print(
    df.groupby("PRIORITY")
    .size()
)
```

```python
#Group by client name and find the total billable amount.
print(
    df.groupby("PRIORITY")
    .size()
)
```

```python
#Find the top 5 clients by number of tickets.
print(
    df.groupby("Client Name ")["Billable Amount USD"]
    .sum()
)
```

```python
#Find the average CSAT score for each priority.
print(
    df.groupby("PRIORITY")["CSAT Score"]
    .mean()
)
```

```python
#Draw a bar chart of ticket count by service line.
df["ServiceLine"].value_counts().plot(kind="bar")
plt.xlabel("Service Line")
plt.ylabel("Ticket Count")
plt.title("Ticket Count by Service Line")
plt.show()
```

```python
#Draw a histogram of hours logged.
df["ServiceLine"].value_counts().plot(kind="bar")
plt.xlabel("Service Line")
plt.ylabel("Ticket Count")
plt.title("Ticket Count by Service Line")
plt.show()
```

```python
#Show only the tickets where hours logged is more than 50.
print(df[df["hours_logged"] > 50])
```

```python
#Show the tickets where status is Closed and priority is P1.
print(
    df[
        (df["Status"] == "Closed") &
        (df["PRIORITY"] == "P1")
    ]
)
```

```python
#In the SLA breached column, change Yes, Y, TRUE and 1 to True, and everything else to False.
df["SLA Breached"] = (
    df["SLA Breached"]
    .astype(str)
    .str.strip()
    .str.upper()
    .isin(["YES", "Y", "TRUE", "1"])
)

print(df["SLA Breached"])
```

```python
#Save your cleaned table to a new CSV file.
df.to_csv(
    "it_services_tickets_cleaned.csv",
    index=False
)
```
