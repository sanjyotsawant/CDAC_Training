```python
#Make an array from the hours logged column.
hours_array = np.array(df["hours_logged"])
print(hours_array)
```

```python
#Replace all the missing values with 0.
hours_no_nan = np.nan_to_num(
    hours_array,
    nan=0
)

print(hours_no_nan)
```

```python
#Count how many values are greater than 100.
print(np.sum(hours_no_nan > 100))
```

```python
#Show values greater than 100
print(hours_no_nan[hours_no_nan > 100])
```

```python
#Count how many values are negative.
print(np.sum(hours_no_nan < 0))
```

```python
#Replace all the negative values with 0.
hours_positive = hours_no_nan.copy()
hours_positive[hours_positive < 0] = 0

print(hours_positive)
```

```python
#Sort the array from smallest to largest.
sorted_hours = np.sort(hours_positive)
print(sorted_hours)
```

```python
#Multiply the hours array by the billing rate array.
billing_rate_array = pd.to_numeric(
    df["Billing Rate (USD)"],
    errors="coerce"
).to_numpy()

revenue = hours_no_nan * billing_rate_array

print(revenue)
```

```python
#Check whether the two arrays have the same length.
print(len(hours_positive) == len(billing_rate_array))
```
